## EDA

**5. Data Analysis**

**Univariate Analysis**


```python
# Dealing with Outliers
df = data[data['price'] <  1200]
sns.boxplot(data=df, y='price')
```

```python
# Price Distribution

plt.figure(figsize=(8,5))
sns.histplot(data=df,x='price',bins=50)
plt.ylabel('Frequency')
plt.xlabel('Price')
plt.title('Price Distribution')
plt.show()
```

```python
df.columns
```

```python
df.dtypes
```

```python
# Availability_365 Distribution

plt.figure(figsize=(8,5))
sns.histplot(data=df,x='availability_365')
plt.ylabel('Frequency')
plt.xlabel('Availability_365')
plt.title('Availability_365 Distribution')
plt.show()
```

```python
# Group By 

df.groupby(by='neighbourhood_group')['price'].mean()
```

```python
# Average Price per Bed 

df.groupby(by='neighbourhood_group')['price per bed'].mean()
```

**Feature Engineering**

```python
df.head()
df['price per bed'] = df['price']/df['beds']
df.head()
```

**Bivariate Analysis**
*One Variable Dependent on Another Variable*

```python
df.columns
```

```python
# Price across Neighbourhood Group

sns.barplot(data=df,x='neighbourhood_group',y='price',hue='room_type')
plt.ylabel('Price')
plt.xlabel('Neighbourhood Group')
plt.title('Price across Neighbourhood Group')
plt.show()
```

```python
# Reviews Based on Price

sns.scatterplot(data=df,x='number_of_reviews',y='price',hue='neighbourhood_group')
plt.ylabel('Price')
plt.xlabel('Reviews')
plt.title('Reviews Based on Price')
plt.show()
```

```python
df.dtypes
```

```python
# Pair Plot

sns.pairplot(data=df,vars=['price','minimum_nights','number_of_reviews','availability_365',],hue='room_type')
```

```python
# Scatter Plot 

sns.scatterplot(data=df, x='longitude',y='latitude',hue='room_type')
plt.ylabel('Latitude')
plt.xlabel('Longitude')
plt.title('Geographical Distribution of AirBnB Listing')
plt.show()
```

```python
df.dtypes
```

```python
# Heat Map

corr = df[['latitude','longitude','price','minimum_nights','number_of_reviews','reviews_per_month','availability_365','beds']].corr()
corr

sns.heatmap(data=corr,annot=True)
```
